```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Student_ID': range(1, 11),
    'Math_Score': [78, 85, np.nan, 65, 99, 45, 88, 92, 110, 60],
    'Science_Score': [82, 79, 88, 94, np.nan, 50, 95, 101, 77, 65],
    'English_Score': [70, 75, 80, 60, 90, 55, 85, 100, 95, np.nan],
    'Attendance': [85, 90, 80, 70, 95, 50, 88, 100, 105, 65]  # Notice 105% attendance (error)
}

df = pd.DataFrame(data)
df.head()
```

```python
df.isnull().sum()
```

```python
df['Math_Score'].fillna(df['Math_Score'].mean(),inplace=True)
df['Science_Score'].fillna(df['Science_Score'].mean(),inplace=True)
df['English_Score'].fillna(df['English_Score'].mean(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Attendance'] = df['Attendance'].apply(lambda x: 100 if x > 100 else x)
```

```python
def detect_outliers(series):
    Q1=series.quantile(0.25)
    Q3=series.quantile(0.75)
    IQR=Q3 - Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    return series[(series < lower_bound) | (series > upper_bound)]

numeric_cols = ['Math_Score', 'Science_Score', 'English_Score', 'Attendance']
for col in numeric_cols:
    outliers = detect_outliers(df[col])
    print(f"Outliers in {col}:")
    print(outliers)
```

```python
# Cap scores at 100
df['Math_Score'] = df['Math_Score'].apply(lambda x: 100 if x > 100 else x)
df['Science_Score'] = df['Science_Score'].apply(lambda x: 100 if x > 100 else x)
```
