```python
# Boxplot for each feature
plt.figure(figsize=(14,8))

# Numeric columns
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

for i, feature in enumerate(features, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=df[feature], color='lightgreen')
    plt.title(f'Boxplot of {feature}')

plt.tight_layout()
plt.suptitle('Boxplots of Iris Dataset Features', fontsize=18)
plt.subplots_adjust(top=0.92)
plt.show()
```

```python
import org.apache.spark.sql.SparkSession

val spark = SparkSession.builder().appName("PositiveNegativeZeroCheck").master("local[*]").getOrCreate()

val numbers = List(10, -5, 0, 25, -100, 7)

val numbersRDD = spark.sparkContext.parallelize(numbers)

val resultRDD = numbersRDD.map { num =>
  if (num > 0) s"$num is Positive"
  else if (num < 0) s"$num is Negative"
  else s"$num is Zero"
}

resultRDD.collect().foreach(println)

spark.stop()
```
